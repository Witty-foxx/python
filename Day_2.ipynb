# Basic List Manipulation Techniques

### Adding Elements
- `append()`
- `extend()`
- `insert()`

### Removing Elements
- `remove()`
- `pop()`
- `clear()`

### Reordering Elements
- `sort()`
- `reverse()`

```python
print("Adding or appending the elements:")
a = [1,2,3,4,5,6,7,8,9]
print(a)
a.append(10)
print(a)
a.extend([11,12,13,14])
print(a)
print(" ")
print("Elimanating the elements:")
a.remove(1)
print(a)
a.pop()
print(a)
print(" ")
print("Sorting and revese sorting:")
arr1 = sorted(a)
print(arr1)
arr2 = sorted(a,reverse=True)
print(arr2)
```

```python
if 1 in arr2:
  print(True)
else:
  print(False)
print()
if 2 in arr2:
  print(True)
else:
  print(False)
```
