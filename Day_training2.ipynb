# 1.Basic Traversal and Aggregation

Given the following array:

`arr = [4, 7, 2, 9, 5, 12]`

## Objective

Traverse the array and find:

1. **Sum of all elements**
2. **Largest element**
3. **Smallest element**
4. **Number of even elements**

## Constraints

* Do **not** use built-in `sum()`, `max()`, or `min()` functions.
* Solve the problem using basic array traversal.

## Concepts Covered

* List creation
* Indexing
* `for` loop
* Array traversal
* Conditional statements
* Variables and accumulators
* Finding minimum and maximum values
* Counting elements based on a condition

## Expected Output

| Operation               | Result |
| ----------------------- | -----: |
| Sum of all elements     |     39 |
| Largest element         |     12 |
| Smallest element        |      2 |
| Number of even elements |      3 |

## Key Learning

This problem introduces the fundamental pattern of **traversal + aggregation**, where each element is processed once while maintaining variables that store the required results.

The same pattern is widely used in problems involving:

* Sum and average
* Minimum and maximum
* Counting elements
* Frequency analysis
* Conditional aggregation

## Complexity

* **Time Complexity:** `O(n)`
* **Space Complexity:** `O(1)`

The array is traversed exactly once, making this an efficient linear-time solution.


````python
arr = [4,7,2,9,5,12]

maxi = 0
mini = 100000
summ = 0
even_num = 0

c = len(arr)

for i in range(c):
  if(arr[i]>maxi):
    maxi = arr[i]
  if(arr[i] < mini):
    mini = arr[i]

  summ=summ+arr[i]

  if(arr[i]%2 == 0):
    even_num=even_num+1



print("The maximum value:",maxi)
print("The minimum value:",mini)
print("The summation:",summ)
print("The count of even nums:",even_num)
````

# 2.Reverse a List

## Problem

Given the following list:

`arr = [1, 2, 3, 4, 5]`

Create a new list containing the elements in reverse order.

## Expected Output

`[5, 4, 3, 2, 1]`

## Rules

Do **not** use:

* `reverse()`
* Slicing: `[::-1]`

The solution should be implemented using basic indexing and traversal.

## Concepts Covered

* Indexing
* `len()`
* List traversal
* Creating a new list
* Working with indices
* Reverse indexing

## Key Learning

This problem demonstrates how to traverse a list **from the last index to the first index** and build a new list using the elements encountered during the traversal.

## Complexity

* **Time Complexity:** `O(n)`
* **Space Complexity:** `O(n)`

The list is traversed once, while an additional list is created to store the reversed elements.


````python
arr =[1,2,3,4,5]
arr1 = []
for i in range(4,-1,-1):
  arr1.append(arr[i])

print(arr1)
````

# 3.Move Zeroes

## Problem

Given the following list:

`arr = [0, 1, 0, 3, 12]`

Move all zeroes to the **end of the list** while maintaining the original order of the non-zero elements.

## Expected Output

`[1, 3, 12, 0, 0]`

## Rules

* Try to **modify the existing list itself**.
* Maintain the relative order of all non-zero elements.
* Use basic list operations and index manipulation.

## Concepts Covered

* List modification
* `append()`
* Index manipulation
* List traversal
* Conditional statements
* In-place modification
* Basic DSA problem-solving

## Key Learning

The important idea is to separate the problem into two operations:

1. Identify and move the non-zero elements toward the beginning.
2. Place the required number of zeroes at the end.

This introduces the concept of **in-place array/list manipulation**, which is a fundamental DSA technique.

## Complexity

* **Time Complexity:** `O(n)`
* **Space Complexity:** `O(1)` for an in-place approach

## Why This Problem Matters

This is an important beginner-level DSA problem because it develops the ability to manipulate an array while **preserving element order** and minimizing additional space.

It is also a good introduction to the **two-pointer technique**, which appears frequently in array and string problems.


````python
arr = [0,1,0,3,12]

c = len(arr)
a = 0

for i in range(1,c):
  if(arr[a] == 0 and arr[i] != 0):
    arr[a],arr[i] = arr[i],arr[a]
    a=a+1

print(arr)
````

# 4.Searching + Frequency

## Problem

Given the following list and target:

`arr = [2, 5, 2, 8, 2, 10, 5]`

`target = 2`

Find:

1. Whether `target` exists in the list.
2. The **first index** where `target` occurs.
3. The **number of times** `target` occurs.

## Expected Output

```text
Exists: True
First index: 0
Count: 3
```

## Rules

Do **not** use:

* `index()`
* `count()`

Solve the problem using basic traversal, conditions, and index tracking.

## Concepts Covered

* Searching
* List traversal
* Index tracking
* Counting occurrences
* Conditional statements
* Boolean variables

## Key Learning

This problem introduces the basic **linear search** pattern while simultaneously tracking multiple pieces of information during a single traversal.


````python
arr = [2, 5, 2, 8, 2, 10, 5]

target = 2
nus = 0

c = len(arr)

for i in range(c):
  if(arr[i] == target):
    nus=nus+1
print(nus)
````

# Two Pointers — Remove Duplicates

## Problem

Given a **sorted** list:

`arr = [1, 1, 2, 2, 3, 3, 4, 5, 5]`

Remove the duplicate elements so that each value appears only once.

## Expected Output

`[1, 2, 3, 4, 5]`

## Rules

Do **not** use:

* `set()`

Try solving the problem using **two indices/pointers**.

## Concepts Covered

* Sorted lists
* Index manipulation
* Two-pointer technique
* Duplicate detection
* In-place thinking
* List modification

## Key Learning

Because the list is sorted, duplicate values appear next to each other. This property allows two pointers to efficiently identify unique elements without using an additional set.

## Why This Problem Matters

This is an important transition from basic Python list operations to **core DSA techniques**. The two-pointer pattern is widely used in array, string, searching, and optimization problems.


````python
arr = [1, 1, 2, 2, 3, 3, 4, 5, 5]

arr = sorted(arr)

arr1 = [arr[0]]

for i in range(1, len(arr)):
    if arr[i] != arr[i - 1]:
        arr1.append(arr[i])

print(arr1)
````
